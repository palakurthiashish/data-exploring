```python
#2.Load the supplied dataset and identify the numerical features that will be used for clustering. Check for missing values and explain whether any data-cleaning action is required before modeling.
import pandas as pd

df = pd.read_csv("Food_Batch_Clustering_Day17_Dataset.csv")

print(df.head())

print(df.isnull().sum())
```

```python
features = ['temperature', 'moisture', 'storage_days', 'ph']

X = df[features]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

X_scaled_df.head()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
df['Cluster'] = cluster_labels

df.head()
```

```python
cluster_counts = df['Cluster'].value_counts().sort_index()

print(cluster_counts)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

cluster_labels = kmeans.fit_predict(X_scaled)

df['Cluster'] = cluster_labels

print("Number of observations in each cluster:")
print(df['Cluster'].value_counts().sort_index())
```

```python
# Add cluster labels to the original dataset
df['cluster'] = kmeans.labels_

# Calculate mean of original numerical features for each cluster
cluster_summary = df.groupby('cluster')[X.columns].mean()

# Display the summary
print(cluster_summary)
```

```python
#8Calculate either K-Means inertia or silhouette score for the 3-cluster solution. Report the value and briefly explain what it tells you about the cluster you about the clustering result.

from sklearn.metrics import silhouette_score

sil_score = silhouette_score(X_scaled, kmeans.labels_)

print("Silhouette Score:", sil_score)
```
